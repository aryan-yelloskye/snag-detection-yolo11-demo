```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"   # must come before importing ultralytics

from ultralytics import YOLO

model = YOLO("yolo11n.pt")

model.train(
    data="dataset/data.yaml",
    imgsz=640,
    epochs=50,
    batch=4,
    device="cpu",
    workers=0,
    cache=False,
    name="train_v2"
)
```

```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

from ultralytics import YOLO

model = YOLO(r"runs\detect\train-13\weights\best.pt")
results = model.predict(
    source=r"testtt.jpg",
    imgsz=640,
    conf=0.25,
    device="cpu",
    save=True
)

# show the result inside the notebook
results[0].show()
```
